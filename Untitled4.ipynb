```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("resultados.csv")

plt.figure(figsize=(10, 6))

# Graficamos la energía cinética total y la individual de cada bola
plt.plot(df['t'], df['Ec_total'], label='Energía Cinética Total', color='purple', linewidth=2.5)
plt.plot(df['t'], df['Ec1'], label='Bola 1 (Ec1)', linestyle='--', color='blue', alpha=0.7)
plt.plot(df['t'], df['Ec2'], label='Bola 2 (Ec2)', linestyle='--', color='orange', alpha=0.7)

plt.title('Comportamiento de la Energía Cinética en la Colisión')
plt.xlabel('Tiempo (s)')
plt.ylabel('Energía Cinética (J)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.7)
plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 6))

# Graficamos las velocidades en X de ambas bolas
plt.plot(df['t'], df['vx1'], label='Velocidad X - Bola 1', color='blue')
plt.plot(df['t'], df['vx2'], label='Velocidad X - Bola 2', color='orange')

plt.title('Velocidades durante la Simulación (Eje X)')
plt.xlabel('Tiempo (s)')
plt.ylabel('Velocidad (m/s)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.7)
plt.tight_layout()
plt.show()
```
